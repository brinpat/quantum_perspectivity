```python
import math


def fibonacci_sphere(samples, amplitude, xc, yc, zc):

    points = []
    phi = np.pi * (3. - np.sqrt(5.))  # golden angle in radians

    for i in range(samples):
        y = amplitude[i]*(1 - (i / float(samples - 1)) * 2)  # y goes from 1 to -1
        radius = np.sqrt(amplitude[i]**2 - y * y)  # radius at y

        theta = phi * i  # golden angle increment

        x = np.cos(theta) * radius
        z = np.sin(theta) * radius

        points.append((x+xc, y+yc, z+zc))

    return np.transpose(points)
```

```python
import matplotlib.animation as animation

fig = plt.figure()
ax = plt.axes(projection='3d')

# mutable particle positions (Nbody, 3) — these get updated frame over frame
p1 = s1.T.copy()
p2 = s2.T.copy()

G = 300.0          # attraction strength — raise for a faster, more violent collapse
softening = 25.0   # prevents force blowup when particles get very close
max_step = 3.0     # per-frame speed cap so it doesn't explode numerically

def frame(w):
    global p1, p2
    ax.clear()

    # pairwise attraction: every particle in sphere 1 is pulled toward every
    # particle in sphere 2 (and vice versa), weighted by combined wave amplitude
    # and inverse-square distance — this is what lets near particles react
    # faster than far ones, instead of the whole sphere moving as one rigid block
    diff = p2[None,:,:] - p1[:,None,:]                 # (N1, N2, 3)
    dist2 = np.sum(diff**2, axis=-1) + softening**2
    dist = np.sqrt(dist2)
    amp_pair = np.abs(A1)[:,None] * np.abs(A2)[None,:]  # combined amplitude weight
    force_mag = G * amp_pair / dist2
    force_dir = diff / dist[..., None]

    disp1 = np.clip((force_mag[...,None]*force_dir).sum(axis=1)/Nbody, -max_step, max_step)
    disp2 = np.clip(-(force_mag[...,None]*force_dir).sum(axis=0)/Nbody, -max_step, max_step)
    p1 += disp1
    p2 += disp2

    ax.plot3D(p1[:,0], p1[:,1], p1[:,2], alpha=0.3, color='r')
    ax.plot3D(p2[:,0], p2[:,1], p2[:,2], alpha=0.3, color='b')
    plot = ax.scatter3D(p1[:,0], p1[:,1], p1[:,2], color='r', s=4)
    ax.scatter3D(p2[:,0], p2[:,1], p2[:,2], color='b', s=4)

    return plot

plt.close()
anim = animation.FuncAnimation(fig, frame, frames=111, blit=False, repeat=True)
anim
```

```python
import matplotlib.animation as animation

fig = plt.figure()
ax = plt.axes(projection='3d')

p1 = s1.T.copy()
p2 = s2.T.copy()
v1 = np.zeros_like(p1)   # NEW: velocity, not just position
v2 = np.zeros_like(p2)

G = 25.0           # lower than before — momentum builds up energy over time, so less force is needed
softening = 25.0
damping = 0.995    # <1.0 bleeds off a little energy each frame so it settles instead of flying apart forever
dt = 1.0

def frame(w):
    global p1, p2, v1, v2
    ax.clear()

    diff = p2[None,:,:] - p1[:,None,:]
    dist2 = np.sum(diff**2, axis=-1) + softening**2
    dist = np.sqrt(dist2)
    amp_pair = np.abs(A1)[:,None] * np.abs(A2)[None,:]
    force_mag = G * amp_pair / dist2
    force_dir = diff / dist[..., None]

    a1 = (force_mag[...,None]*force_dir).sum(axis=1) / Nbody
    a2 = -(force_mag[...,None]*force_dir).sum(axis=0) / Nbody

    v1[:] = (v1 + a1*dt) * damping   # accumulate velocity instead of setting position directly
    v2[:] = (v2 + a2*dt) * damping
    p1 += v1*dt
    p2 += v2*dt

    ax.set_xlim3d([-400,1000])
    ax.set_ylim3d([-400,1000])
    ax.set_zlim3d([-400,400])

    ax.scatter3D(p1[:,0], p1[:,1], p1[:,2], color='r', s=4)
    plot = ax.scatter3D(p2[:,0], p2[:,1], p2[:,2], color='b', s=4)
    return plot

plt.close()
anim = animation.FuncAnimation(fig, frame, frames=150, blit=False, repeat=True)
anim
```
